```python
# ============================================
# STEP 5.1 — CREATE PROCESSED COPIES
# ============================================

import pandas as pd
import numpy as np
import re
import csv
from pathlib import Path

# Processed-data folder
PROCESSED = Path("/content/PACKVOTE/data/processed")
PROCESSED.mkdir(parents=True, exist_ok=True)

# Never modify the original loaded dataframes
users_p = users.copy()
destinations_p = destinations.copy()
reviews_p = reviews.copy()
user_history_p = user_history.copy()
travel_cost_p = travel_cost.copy()
tourist_spots_p = tourist_spots.copy()
characteristics_p = characteristics.copy()

print("✅ Processed copies created")
```

```python
# ============================================
# STEP 5.2 — GENERAL CLEANING
# ============================================

def clean_column_names(df):
    df = df.copy()
    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
        .str.replace(" ", "_", regex=False)
    )
    return df


def clean_text_columns(df):
    df = df.copy()

    for col in df.select_dtypes(include=["object"]).columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
        )

        # Convert string versions of missing values to real NaN
        df[col] = df[col].replace(
            ["", "nan", "None", "NULL", "null"],
            np.nan
        )

    return df


users_p = clean_text_columns(clean_column_names(users_p))
destinations_p = clean_text_columns(clean_column_names(destinations_p))
reviews_p = clean_text_columns(clean_column_names(reviews_p))
user_history_p = clean_text_columns(clean_column_names(user_history_p))
travel_cost_p = clean_text_columns(clean_column_names(travel_cost_p))
tourist_spots_p = clean_text_columns(clean_column_names(tourist_spots_p))
characteristics_p = clean_text_columns(clean_column_names(characteristics_p))

print("✅ Column names and text values cleaned")
```

```python
# ============================================
# STEP 5.3A — USERS
# ============================================

# Fill missing preference values
users_p["Preferences"] = users_p["Preferences"].fillna("Unknown")

# Standardize preference text
def normalize_preferences(text):
    text = str(text).lower().strip()

    mapping = {
        "beaches": "Beach",
        "beach": "Beach",
        "historical": "Historical",
        "history": "Historical",
        "nature": "Nature",
        "adventure": "Adventure",
        "city": "City"
    }

    parts = [p.strip() for p in text.split(",")]

    normalized = []
    for part in parts:
        if part in mapping:
            normalized.append(mapping[part])

    # Remove duplicates while preserving order
    normalized = list(dict.fromkeys(normalized))

    return ", ".join(normalized) if normalized else "Unknown"


users_p["Preferences_Clean"] = (
    users_p["Preferences"]
    .apply(normalize_preferences)
)

# Numeric columns
users_p["NumberOfAdults"] = pd.to_numeric(
    users_p["NumberOfAdults"],
    errors="coerce"
)

users_p["NumberOfChildren"] = pd.to_numeric(
    users_p["NumberOfChildren"],
    errors="coerce"
)

# Fill missing numeric values with median
for col in ["NumberOfAdults", "NumberOfChildren"]:
    users_p[col] = users_p[col].fillna(users_p[col].median())

print("Preference values after cleaning:")
print(users_p["Preferences_Clean"].value_counts())

display(
    users_p[
        ["UserID", "Preferences", "Preferences_Clean",
         "NumberOfAdults", "NumberOfChildren"]
    ].head(10)
)
```

```python
# ============================================
# STEP 5.4 — PREFERENCE FEATURES
# ============================================

preference_categories = [
    "Beach",
    "Historical",
    "Nature",
    "Adventure",
    "City"
]

for category in preference_categories:
    users_p[f"Pref_{category}"] = (
        users_p["Preferences_Clean"]
        .str.contains(category, case=False, na=False)
        .astype(int)
    )

print("✅ Preference features created")

display(
    users_p[
        ["UserID", "Preferences_Clean"] +
        [f"Pref_{c}" for c in preference_categories]
    ].head(10)
)
```

```python
# ============================================
# STEP 5.5 — DESTINATIONS
# ============================================

destinations_p["DestinationID"] = pd.to_numeric(
    destinations_p["DestinationID"],
    errors="coerce"
)

destinations_p["Popularity"] = pd.to_numeric(
    destinations_p["Popularity"],
    errors="coerce"
)

destinations_p["Name"] = destinations_p["Name"].fillna("Unknown")
destinations_p["State"] = destinations_p["State"].fillna("Unknown")
destinations_p["Type"] = destinations_p["Type"].fillna("Unknown")
destinations_p["BestTimeToVisit"] = (
    destinations_p["BestTimeToVisit"]
    .fillna("Unknown")
)

# Remove duplicate destination records only in processed data
destinations_p = destinations_p.drop_duplicates()

print("✅ Destinations cleaned")
display(destinations_p.head())
```

```python
# ============================================
# STEP 5.6 — REVIEWS
# ============================================

reviews_p["Rating"] = pd.to_numeric(
    reviews_p["Rating"],
    errors="coerce"
)

reviews_p["ReviewText"] = (
    reviews_p["ReviewText"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace(r"[^a-zA-Z0-9\s]", " ", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Flag ratings outside the expected 1–5 range
reviews_p["Invalid_Rating"] = (
    (reviews_p["Rating"] < 1) |
    (reviews_p["Rating"] > 5)
)

# Keep a valid-rating version for ML
reviews_ml = reviews_p[
    reviews_p["Rating"].between(1, 5)
].copy()

print("Total reviews:", len(reviews_p))
print("Valid ratings:", len(reviews_ml))
print("Invalid ratings:", reviews_p["Invalid_Rating"].sum())

display(reviews_p.head())
```

```python
# ============================================
# STEP 5.7 — USER HISTORY
# ============================================

user_history_p["HistoryID"] = pd.to_numeric(
    user_history_p["HistoryID"],
    errors="coerce"
)

user_history_p["UserID"] = pd.to_numeric(
    user_history_p["UserID"],
    errors="coerce"
)

user_history_p["DestinationID"] = pd.to_numeric(
    user_history_p["DestinationID"],
    errors="coerce"
)

user_history_p["ExperienceRating"] = pd.to_numeric(
    user_history_p["ExperienceRating"],
    errors="coerce"
)

user_history_p["VisitDate"] = pd.to_datetime(
    user_history_p["VisitDate"],
    errors="coerce"
)

print("✅ User history cleaned")
display(user_history_p.head())
```

```python
# ============================================
# STEP 5.8 — ACCOMMODATION COST
# ============================================

def parse_cost(value):
    if pd.isna(value):
        return np.nan, np.nan, np.nan, "MISSING"

    text = str(value).strip()

    numbers = re.findall(
        r"\d+(?:\.\d+)?",
        text
    )

    if len(numbers) == 0:
        return np.nan, np.nan, np.nan, "INVALID"

    if len(numbers) == 1:
        value1 = float(numbers[0])
        return value1, value1, value1, "SINGLE_VALUE"

    value1 = float(numbers[0])
    value2 = float(numbers[1])

    if value1 > value2:
        return (
            value2,
            value1,
            (value1 + value2) / 2,
            "INVERTED_CORRECTED"
        )

    return (
        value1,
        value2,
        (value1 + value2) / 2,
        "VALID"
    )


parsed = travel_cost_p["Accomdation_Cost"].apply(parse_cost)

travel_cost_p[
    ["Cost_Min", "Cost_Max", "Cost_Average", "Cost_Flag"]
] = pd.DataFrame(
    parsed.tolist(),
    index=travel_cost_p.index
)

print("Cost flag distribution:")
print(travel_cost_p["Cost_Flag"].value_counts())

display(
    travel_cost_p[
        [
            "City",
            "Accomadation_Type",
            "Accomdation_Cost",
            "Cost_Min",
            "Cost_Max",
            "Cost_Average",
            "Cost_Flag"
        ]
    ].head(20)
)
```

```python
# ============================================
# STEP 5.9 — TOURIST SPOTS
# ============================================

tourist_spots_p["DestinationID"] = pd.to_numeric(
    tourist_spots_p["DestinationID"],
    errors="coerce"
)

tourist_spots_p["Latitude"] = pd.to_numeric(
    tourist_spots_p["Latitude"],
    errors="coerce"
)

tourist_spots_p["Longitude"] = pd.to_numeric(
    tourist_spots_p["Longitude"],
    errors="coerce"
)

tourist_spots_p["Name"] = (
    tourist_spots_p["Name"]
    .fillna("Unknown")
)

tourist_spots_p["Characteristics"] = (
    tourist_spots_p["Characteristics"]
    .fillna("Unknown")
    .astype(str)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

tourist_spots_p = tourist_spots_p.drop_duplicates()

print("✅ Tourist spots cleaned")
display(tourist_spots_p.head())
```

```python
# ============================================
# STEP 5.10 — CHARACTERISTICS
# ============================================

characteristics_p["characteristics"] = (
    characteristics_p["characteristics"]
    .fillna("Unknown")
    .astype(str)
    .str.lower()
    .str.strip()
)

characteristics_p = characteristics_p.drop_duplicates()

print("✅ Characteristics cleaned")
display(characteristics_p.head(10))
```

```python
# ============================================
# STEP 5.11 — USER VISITS
# ============================================

def read_user_visits(path):

    rows = []

    with open(
        path,
        "r",
        encoding="utf-8",
        errors="replace"
    ) as file:

        for line in file:
            line = line.rstrip("\n")

            # Remove outer quotation
            if line.startswith('"') and line.endswith('"'):
                line = line[1:-1]

            # Restore doubled quotes
            line = line.replace('""', '"')

            row = next(
                csv.reader(
                    [line],
                    delimiter=";"
                )
            )

            rows.append(row)

    return pd.DataFrame(
        rows[1:],
        columns=rows[0]
    )


user_visits_p = read_user_visits(user_visits_path)

# Convert numeric columns
for col in ["photoID", "userID", "poiID", "poiFreq", "seqID"]:
    user_visits_p[col] = pd.to_numeric(
        user_visits_p[col],
        errors="coerce"
    )

user_visits_p["dateTaken"] = pd.to_datetime(
    user_visits_p["dateTaken"],
    errors="coerce"
)

print("User Visits shape:", user_visits_p.shape)
print("Columns:")
print(user_visits_p.columns.tolist())

display(user_visits_p.head())
```

```python
# ============================================================
# STEP 6 — PACKVOTE FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

# ------------------------------------------------------------
# 1. TRAVELER FEATURES
# ------------------------------------------------------------

experience_features = (
    user_history_p.groupby("UserID")
    .agg(
        Avg_Experience=("ExperienceRating", "mean"),
        Visit_Count=("DestinationID", "count")
    )
    .reset_index()
)

review_features = (
    reviews_ml.groupby("UserID")
    .agg(
        Avg_Review_Rating=("Rating", "mean"),
        Review_Count=("ReviewID", "count")
    )
    .reset_index()
)

traveler_features = users_p[
    [
        "UserID",
        "NumberOfAdults",
        "NumberOfChildren",
        "Pref_Beach",
        "Pref_Historical",
        "Pref_Nature",
        "Pref_Adventure",
        "Pref_City"
    ]
].copy()

traveler_features["Group_Size"] = (
    traveler_features["NumberOfAdults"] +
    traveler_features["NumberOfChildren"]
)

traveler_features = (
    traveler_features
    .merge(experience_features, on="UserID", how="left")
    .merge(review_features, on="UserID", how="left")
)

for col in [
    "Avg_Experience",
    "Visit_Count",
    "Avg_Review_Rating",
    "Review_Count"
]:
    traveler_features[col] = traveler_features[col].fillna(0)


# ------------------------------------------------------------
# 2. DESTINATION FEATURES
# ------------------------------------------------------------

destination_features = destinations_p[
    [
        "DestinationID",
        "Name",
        "State",
        "Type",
        "Popularity",
        "BestTimeToVisit"
    ]
].copy()

destination_type_features = pd.get_dummies(
    destination_features["Type"],
    prefix="Type",
    dtype=int
)

destination_features = pd.concat(
    [destination_features, destination_type_features],
    axis=1
)

pop_min = destination_features["Popularity"].min()
pop_max = destination_features["Popularity"].max()

destination_features["Popularity_Norm"] = (
    destination_features["Popularity"] - pop_min
) / (pop_max - pop_min + 1e-9)


# ------------------------------------------------------------
# 3. DESTINATION RATING FEATURES
# ------------------------------------------------------------

destination_rating_features = (
    reviews_ml
    .merge(
        destinations_p[["DestinationID", "Name"]],
        on="DestinationID",
        how="left"
    )
    .groupby(["DestinationID", "Name"])
    .agg(
        Avg_Rating=("Rating", "mean"),
        Rating_Count=("ReviewID", "count")
    )
    .reset_index()
)

destination_rating_features["Rating_Norm"] = (
    destination_rating_features["Avg_Rating"] / 5
)

destination_features = destination_features.merge(
    destination_rating_features,
    on=["DestinationID", "Name"],
    how="left"
)

destination_features[
    ["Avg_Rating", "Rating_Count", "Rating_Norm"]
] = destination_features[
    ["Avg_Rating", "Rating_Count", "Rating_Norm"]
].fillna(0)


# ------------------------------------------------------------
# 4. USER + DESTINATION INTERACTION FEATURES
# ------------------------------------------------------------

interaction_features = (
    user_history_p[
        [
            "UserID",
            "DestinationID",
            "ExperienceRating",
            "VisitDate"
        ]
    ]
    .merge(
        users_p[
            [
                "UserID",
                "Preferences_Clean",
                "NumberOfAdults",
                "NumberOfChildren"
            ]
        ],
        on="UserID",
        how="left"
    )
    .merge(
        destinations_p[
            [
                "DestinationID",
                "Name",
                "State",
                "Type",
                "Popularity"
            ]
        ],
        on="DestinationID",
        how="left"
    )
)

interaction_features["Group_Size"] = (
    interaction_features["NumberOfAdults"] +
    interaction_features["NumberOfChildren"]
)

interaction_features["Preference_Match"] = (
    interaction_features.apply(
        lambda row: int(
            str(row["Type"]).lower()
            in str(row["Preferences_Clean"]).lower()
        ),
        axis=1
    )
)

# Target for destination suitability
interaction_features["Suitable"] = (
    interaction_features["ExperienceRating"] >= 4
).astype(int)


# ------------------------------------------------------------
# 5. KNN FEATURES
# ------------------------------------------------------------

interaction_preferences = interaction_features[
    "Preferences_Clean"
].apply(
    lambda text: pd.Series({
        "Pref_Beach": int("beach" in str(text).lower()),
        "Pref_Historical": int("historical" in str(text).lower()),
        "Pref_Nature": int("nature" in str(text).lower()),
        "Pref_Adventure": int("adventure" in str(text).lower()),
        "Pref_City": int("city" in str(text).lower())
    })
)

interaction_type = pd.get_dummies(
    interaction_features["Type"],
    prefix="Type",
    dtype=int
)

knn_features = pd.concat(
    [
        interaction_features[
            [
                "NumberOfAdults",
                "NumberOfChildren",
                "Group_Size",
                "Popularity",
                "Preference_Match",
                "ExperienceRating"
            ]
        ].reset_index(drop=True),
        interaction_preferences.reset_index(drop=True),
        interaction_type.reset_index(drop=True)
    ],
    axis=1
)

knn_target = interaction_features["ExperienceRating"].astype(float)


# ------------------------------------------------------------
# 6. NAIVE BAYES / REVIEW TEXT FEATURES
# ------------------------------------------------------------

reviews_nb = reviews_ml.copy()

reviews_nb["Review_Label"] = (
    reviews_nb["Rating"] >= 4
).astype(int)

tfidf_vectorizer = TfidfVectorizer(
    max_features=1000,
    ngram_range=(1, 2),
    stop_words="english"
)

X_review_text = tfidf_vectorizer.fit_transform(
    reviews_nb["ReviewText"].astype(str)
)

y_review_text = reviews_nb["Review_Label"]


# ------------------------------------------------------------
# 7. ACCOMMODATION COST FEATURES
# ------------------------------------------------------------

cost_features = travel_cost_p[
    [
        "City",
        "Accomadation_Type",
        "Cost_Min",
        "Cost_Max",
        "Cost_Average"
    ]
].copy()

cost_features = cost_features[
    cost_features["Cost_Average"].notna()
].copy()

cost_encoded = pd.get_dummies(
    cost_features[
        ["City", "Accomadation_Type"]
    ],
    columns=["City", "Accomadation_Type"],
    dtype=int
)

cost_ml_features = pd.concat(
    [
        cost_encoded.reset_index(drop=True),
        cost_features[
            ["Cost_Min", "Cost_Max"]
        ].reset_index(drop=True)
    ],
    axis=1
)

cost_target = cost_features["Cost_Average"].astype(float)


# ------------------------------------------------------------
# 8. TOURIST SPOT FEATURES
# ------------------------------------------------------------

tourist_features = tourist_spots_p.copy()

tourist_features["Characteristic_Count"] = (
    tourist_features["Characteristics"]
    .fillna("")
    .apply(
        lambda x: len([
            item for item in str(x).split(",")
            if item.strip()
        ])
    )
)

tourist_features["Has_Nature"] = (
    tourist_features["Characteristics"]
    .str.contains(
        "nature|scenic|greenery|waterfall",
        case=False,
        na=False
    )
).astype(int)

tourist_features["Has_Adventure"] = (
    tourist_features["Characteristics"]
    .str.contains(
        "adventure|trekking|hiking|zipline",
        case=False,
        na=False
    )
).astype(int)

tourist_features["Has_Historical"] = (
    tourist_features["Characteristics"]
    .str.contains(
        "historical|heritage|temple|fort",
        case=False,
        na=False
    )
).astype(int)

tourist_features["Has_Family"] = (
    tourist_features["Characteristics"]
    .str.contains(
        "family|amusement",
        case=False,
        na=False
    )
).astype(int)

tourist_summary = (
    tourist_features
    .groupby("DestinationID")
    .agg(
        Tourist_Spot_Count=("Name", "count"),
        Nature_Spot_Count=("Has_Nature", "sum"),
        Adventure_Spot_Count=("Has_Adventure", "sum"),
        Historical_Spot_Count=("Has_Historical", "sum"),
        Family_Spot_Count=("Has_Family", "sum")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 9. SAVE FEATURE TABLES
# ------------------------------------------------------------

traveler_features.to_csv(
    PROCESSED / "traveler_features.csv",
    index=False
)

destination_features.to_csv(
    PROCESSED / "destination_features.csv",
    index=False
)

interaction_features.to_csv(
    PROCESSED / "interaction_features.csv",
    index=False
)

knn_features.to_csv(
    PROCESSED / "knn_features.csv",
    index=False
)

cost_ml_features.to_csv(
    PROCESSED / "cost_ml_features.csv",
    index=False
)

tourist_features.to_csv(
    PROCESSED / "tourist_features.csv",
    index=False
)

tourist_summary.to_csv(
    PROCESSED / "tourist_summary.csv",
    index=False
)


# ------------------------------------------------------------
# 10. SUMMARY
# ------------------------------------------------------------

print("✅ STEP 6 COMPLETED — FEATURE ENGINEERING")
print()
print("Traveler features      :", traveler_features.shape)
print("Destination features   :", destination_features.shape)
print("Interaction features   :", interaction_features.shape)
print("KNN feature matrix      :", knn_features.shape)
print("TF-IDF review matrix    :", X_review_text.shape)
print("Cost feature matrix     :", cost_ml_features.shape)
print("Tourist features        :", tourist_features.shape)
print("Tourist summary         :", tourist_summary.shape)
```

### STEP 7 — K-MEANS TRAVELER PROFILING

Now that we have prepared all the features, we can use K-Means clustering to create traveler profiles based on their characteristics and preferences. This will help group similar users together.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# ------------------------------------------------------------
# 1. Prepare data for K-Means
# ------------------------------------------------------------

kmeans_data = traveler_features[
    [
        'NumberOfAdults',
        'NumberOfChildren',
        'Group_Size',
        'Avg_Experience',
        'Visit_Count',
        'Avg_Review_Rating',
        'Review_Count',
        'Pref_Beach',
        'Pref_Historical',
        'Pref_Nature',
        'Pref_Adventure',
        'Pref_City'
    ]
].copy()

# Handle potential NaN values (e.g., from users with no history/reviews)
kmeans_data = kmeans_data.fillna(0)

# ------------------------------------------------------------
# 2. Scale features
# ------------------------------------------------------------

kmeans_scaler = StandardScaler()
kmeans_scaled = kmeans_scaler.fit_transform(kmeans_data)

# ------------------------------------------------------------
# 3. Determine optimal number of clusters (Elbow Method)
# ------------------------------------------------------------

inertia = []
k_range = range(1, 11)  # Test K from 1 to 10

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(kmeans_scaled)
    inertia.append(kmeans.inertia_)

# Plot the Elbow Method results
fig = plt.figure(figsize=(10, 6))
plt.plot(k_range, inertia, marker='o')
plt.xlabel('Number of clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.grid(True)
plt.show()

print("\n✅ Elbow Method plot generated. Examine the plot to choose K.")
```

```python
# ------------------------------------------------------------
# 4. Choose optimal K and train K-Means model
#    (Based on the Elbow Method plot, let's assume K=4 for this example)
# ------------------------------------------------------------

# You might need to adjust 'n_clusters' based on the Elbow Method plot above.
optimal_k = 4 # Adjust this value if the elbow plot suggests a different K

kmeans_model = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
traveler_features['Traveler_Cluster'] = kmeans_model.fit_predict(kmeans_scaled)

# ------------------------------------------------------------
# 5. Create cluster profiles
# ------------------------------------------------------------

cluster_profile = (
    traveler_features.groupby('Traveler_Cluster')
    [[
        'NumberOfAdults',
        'NumberOfChildren',
        'Group_Size',
        'Avg_Experience',
        'Visit_Count',
        'Avg_Review_Rating',
        'Review_Count',
        'Pref_Beach',
        'Pref_Historical',
        'Pref_Nature',
        'Pref_Adventure',
        'Pref_City'
    ]]
    .mean()
)

# Save feature names used for clustering
cluster_columns = kmeans_data.columns.tolist()

print(f"✅ K-Means clustering completed with {optimal_k} clusters.")
print("Cluster Profiles:")
display(cluster_profile.round(2))

print("\nTraveler Features with assigned clusters:")
display(traveler_features[['UserID', 'Traveler_Cluster'] + cluster_columns].head())
```

```python
# ============================================================
# STEP 8 — KNN RECOMMENDATION / EXPERIENCE MODEL
# ============================================================

import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 1. Prepare KNN data
# ------------------------------------------------------------

X = knn_features.copy()
y = knn_target.copy()

# Safety cleaning
X = X.replace([np.inf, -np.inf], np.nan)
X = X.fillna(0)

y = pd.to_numeric(y, errors="coerce")

valid_rows = y.notna()

X = X.loc[valid_rows].reset_index(drop=True)
y = y.loc[valid_rows].reset_index(drop=True)

print("KNN input shape :", X.shape)
print("Target shape    :", y.shape)

# ------------------------------------------------------------
# 2. Train-test split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ------------------------------------------------------------
# 3. Feature scaling
# ------------------------------------------------------------

knn_scaler = StandardScaler()

X_train_scaled = knn_scaler.fit_transform(X_train)
X_test_scaled = knn_scaler.transform(X_test)

# ------------------------------------------------------------
# 4. Find a suitable K
# ------------------------------------------------------------

candidate_k = [3, 5, 7, 9, 11, 15]

results = []

for k in candidate_k:

    model = KNeighborsRegressor(
        n_neighbors=k,
        weights="distance"
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    pred = model.predict(
        X_test_scaled
    )

    mae = mean_absolute_error(
        y_test,
        pred
    )

    rmse = mean_squared_error(
        y_test,
        pred
    ) ** 0.5

    r2 = r2_score(
        y_test,
        pred
    )

    results.append({
        "K": k,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

knn_results = pd.DataFrame(results)

print("\nKNN MODEL COMPARISON")
print("=" * 50)
display(knn_results.round(4))

# ------------------------------------------------------------
# 5. Select best K
# ------------------------------------------------------------

best_k_row = knn_results.loc[
    knn_results["MAE"].idxmin()
]

best_k = int(
    best_k_row["K"]
)

print(
    f"✅ Selected K = {best_k}"
)

# ------------------------------------------------------------
# 6. Train final KNN model
# ------------------------------------------------------------

knn_model = KNeighborsRegressor(
    n_neighbors=best_k,
    weights="distance"
)

knn_model.fit(
    X_train_scaled,
    y_train
)

# ------------------------------------------------------------
# 7. Final predictions
# ------------------------------------------------------------

knn_predictions = knn_model.predict(
    X_test_scaled
)

knn_mae = mean_absolute_error(
    y_test,
    knn_predictions
)

knn_rmse = mean_squared_error(
    y_test,
    knn_predictions
) ** 0.5

knn_r2 = r2_score(
    y_test,
    knn_predictions
)

print("\nFINAL KNN RESULTS")
print("=" * 50)
print(f"Best K       : {best_k}")
print(f"MAE          : {knn_mae:.4f}")
print(f"RMSE         : {knn_rmse:.4f}")
print(f"R²           : {knn_r2:.4f}")

# ------------------------------------------------------------
# 8. Save model
# ------------------------------------------------------------

MODELS = Path("/content/PACKVOTE/models")
MODELS.mkdir(
    parents=True,
    exist_ok=True
)

joblib.dump(
    knn_model,
    MODELS / "knn_recommendation_model.pkl"
)

joblib.dump(
    knn_scaler,
    MODELS / "knn_recommendation_scaler.pkl"
)

joblib.dump(
    X.columns.tolist(),
    MODELS / "knn_feature_columns.pkl"
)

knn_results.to_csv(
    PROCESSED / "knn_results.csv",
    index=False
)

print("\n✅ KNN model saved successfully")
print(
    "Model:",
    MODELS / "knn_recommendation_model.pkl"
)

print(
    "Scaler:",
    MODELS / "knn_recommendation_scaler.pkl"
)

print(
    "\n✅ STEP 8 COMPLETED — KNN RECOMMENDATION MODEL"
)
```

```python
# ============================================================
# STEP 9 — DECISION TREE DESTINATION SUITABILITY
# ============================================================

import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ------------------------------------------------------------
# 1. Create ML features WITHOUT target leakage
# ------------------------------------------------------------

dt_base = interaction_features.copy()

# Traveler preference features
dt_pref = dt_base["Preferences_Clean"].apply(
    lambda text: pd.Series({
        "Pref_Beach": int("beach" in str(text).lower()),
        "Pref_Historical": int("historical" in str(text).lower()),
        "Pref_Nature": int("nature" in str(text).lower()),
        "Pref_Adventure": int("adventure" in str(text).lower()),
        "Pref_City": int("city" in str(text).lower())
    })
)

# Destination type one-hot encoding
dt_type = pd.get_dummies(
    dt_base["Type"],
    prefix="Type",
    dtype=int
)

# Numeric predictors
dt_numeric = dt_base[
    [
        "NumberOfAdults",
        "NumberOfChildren",
        "Group_Size",
        "Popularity",
        "Preference_Match"
    ]
].copy()

# Combine features
X_dt = pd.concat(
    [
        dt_numeric.reset_index(drop=True),
        dt_pref.reset_index(drop=True),
        dt_type.reset_index(drop=True)
    ],
    axis=1
)

# Target
y_dt = dt_base["Suitable"].astype(int)

# Safety cleaning
X_dt = X_dt.replace(
    [np.inf, -np.inf],
    np.nan
).fillna(0)

print("Decision Tree feature matrix:", X_dt.shape)
print("Target distribution:")
print(y_dt.value_counts())

# ------------------------------------------------------------
# 2. Train-test split
# ------------------------------------------------------------

X_train_dt, X_test_dt, y_train_dt, y_test_dt = train_test_split(
    X_dt,
    y_dt,
    test_size=0.20,
    random_state=42,
    stratify=y_dt
)

# ------------------------------------------------------------
# 3. Compare different tree depths
# ------------------------------------------------------------

depths = [2, 3, 4, 5, 6, 8, 10]

dt_results = []

for depth in depths:

    model = DecisionTreeClassifier(
        max_depth=depth,
        min_samples_split=10,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42
    )

    model.fit(
        X_train_dt,
        y_train_dt
    )

    pred = model.predict(
        X_test_dt
    )

    dt_results.append({
        "Max_Depth": depth,
        "Accuracy": accuracy_score(y_test_dt, pred),
        "Precision": precision_score(
            y_test_dt,
            pred,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test_dt,
            pred,
            zero_division=0
        ),
        "F1": f1_score(
            y_test_dt,
            pred,
            zero_division=0
        )
    })

dt_results = pd.DataFrame(dt_results)

print("\nDECISION TREE HYPERPARAMETER COMPARISON")
print("=" * 60)
display(
    dt_results.round(4)
)

# ------------------------------------------------------------
# 4. Select best depth using F1 score
# ------------------------------------------------------------

best_row = dt_results.loc[
    dt_results["F1"].idxmax()
]

best_depth = int(
    best_row["Max_Depth"]
)

print(
    f"\n✅ Selected max_depth = {best_depth}"
)

# ------------------------------------------------------------
# 5. Train final Decision Tree
# ------------------------------------------------------------

decision_tree_model = DecisionTreeClassifier(
    max_depth=best_depth,
    min_samples_split=10,
    min_samples_leaf=5,
    class_weight="balanced",
    random_state=42
)

decision_tree_model.fit(
    X_train_dt,
    y_train_dt
)

# ------------------------------------------------------------
# 6. Final evaluation
# ------------------------------------------------------------

dt_pred = decision_tree_model.predict(
    X_test_dt
)

dt_accuracy = accuracy_score(
    y_test_dt,
    dt_pred
)

dt_precision = precision_score(
    y_test_dt,
    dt_pred,
    zero_division=0
)

dt_recall = recall_score(
    y_test_dt,
    dt_pred,
    zero_division=0
)

dt_f1 = f1_score(
    y_test_dt,
    dt_pred,
    zero_division=0
)

print("\nFINAL DECISION TREE RESULTS")
print("=" * 60)
print(f"Max Depth  : {best_depth}")
print(f"Accuracy   : {dt_accuracy:.4f}")
print(f"Precision  : {dt_precision:.4f}")
print(f"Recall     : {dt_recall:.4f}")
print(f"F1 Score   : {dt_f1:.4f}")

# ------------------------------------------------------------
# 7. Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test_dt,
    dt_pred
)

print("\nCONFUSION MATRIX")
print("=" * 60)
print(cm)

print("\nCLASSIFICATION REPORT")
print("=" * 60)
print(
    classification_report(
        y_test_dt,
        dt_pred,
        target_names=[
            "Not Suitable",
            "Suitable"
        ],
        zero_division=0
    )
)

# ------------------------------------------------------------
# 8. Feature importance
# ------------------------------------------------------------

feature_importance = pd.DataFrame({
    "Feature": X_dt.columns,
    "Importance": decision_tree_model.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

print("\nTOP FEATURE IMPORTANCES")
print("=" * 60)
display(
    feature_importance.head(15).round(4)
)

# ------------------------------------------------------------
# 9. Save model and results
# ------------------------------------------------------------

MODELS = Path("/content/PACKVOTE/models")
PROCESSED = Path("/content/PACKVOTE/data/processed")

MODELS.mkdir(
    parents=True,
    exist_ok=True
)

PROCESSED.mkdir(
    parents=True,
    exist_ok=True
)

joblib.dump(
    decision_tree_model,
    MODELS / "decision_tree_suitability.pkl"
)

joblib.dump(
    X_dt.columns.tolist(),
    MODELS / "decision_tree_feature_columns.pkl"
)

dt_results.to_csv(
    PROCESSED / "decision_tree_results.csv",
    index=False
)

feature_importance.to_csv(
    PROCESSED / "decision_tree_feature_importance.csv",
    index=False
)

print("\n✅ Decision Tree model saved successfully")
print(
    "Model:",
    MODELS / "decision_tree_suitability.pkl"
)

print(
    "\n✅ STEP 9 COMPLETED — DECISION TREE DESTINATION SUITABILITY"
)
```

```python
# ============================================================
# STEP 10 — NAIVE BAYES REVIEW CLASSIFICATION
# ============================================================

import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ------------------------------------------------------------
# 1. Prepare review dataset
# ------------------------------------------------------------

nb_data = reviews_ml[
    ["ReviewID", "ReviewText", "Rating"]
].copy()

# Remove empty review text
nb_data["ReviewText"] = (
    nb_data["ReviewText"]
    .fillna("")
    .astype(str)
    .str.strip()
)

nb_data = nb_data[
    nb_data["ReviewText"] != ""
].copy()

# Create classification label
# Rating >= 4 => Positive
# Rating < 4  => Negative
nb_data["Review_Label"] = (
    nb_data["Rating"] >= 4
).astype(int)

print("Review dataset size:", nb_data.shape)

print("\nClass distribution:")
print(
    nb_data["Review_Label"]
    .value_counts()
    .rename(
        index={
            0: "Negative",
            1: "Positive"
        }
    )
)

# ------------------------------------------------------------
# 2. Train-test split
# ------------------------------------------------------------

X_text_train, X_text_test, y_text_train, y_text_test = (
    train_test_split(
        nb_data["ReviewText"],
        nb_data["Review_Label"],
        test_size=0.20,
        random_state=42,
        stratify=nb_data["Review_Label"]
    )
)

# ------------------------------------------------------------
# 3. TF-IDF feature extraction
# ------------------------------------------------------------

nb_vectorizer = TfidfVectorizer(
    max_features=1000,
    ngram_range=(1, 2),
    min_df=1,
    stop_words="english"
)

X_train_nb = nb_vectorizer.fit_transform(
    X_text_train
)

X_test_nb = nb_vectorizer.transform(
    X_text_test
)

print("\nTF-IDF training matrix:", X_train_nb.shape)
print("TF-IDF testing matrix :", X_test_nb.shape)

# ------------------------------------------------------------
# 4. Compare Naive Bayes smoothing values
# ------------------------------------------------------------

alpha_values = [
    0.1,
    0.5,
    1.0,
    1.5,
    2.0
]

nb_results = []

for alpha in alpha_values:

    model = MultinomialNB(
        alpha=alpha
    )

    model.fit(
        X_train_nb,
        y_text_train
    )

    pred = model.predict(
        X_test_nb
    )

    nb_results.append({
        "Alpha": alpha,
        "Accuracy": accuracy_score(
            y_text_test,
            pred
        ),
        "Precision": precision_score(
            y_text_test,
            pred,
            zero_division=0
        ),
        "Recall": recall_score(
            y_text_test,
            pred,
            zero_division=0
        ),
        "F1": f1_score(
            y_text_test,
            pred,
            zero_division=0
        )
    })

nb_results = pd.DataFrame(
    nb_results
)

print("\nNAIVE BAYES COMPARISON")
print("=" * 60)

display(
    nb_results.round(4)
)

# ------------------------------------------------------------
# 5. Select best alpha using F1
# ------------------------------------------------------------

best_nb_row = nb_results.loc[
    nb_results["F1"].idxmax()
]

best_alpha = float(
    best_nb_row["Alpha"]
)

print(
    f"\n✅ Selected Alpha = {best_alpha}"
)

# ------------------------------------------------------------
# 6. Train final Naive Bayes model
# ------------------------------------------------------------

naive_bayes_model = MultinomialNB(
    alpha=best_alpha
)

naive_bayes_model.fit(
    X_train_nb,
    y_text_train
)

# ------------------------------------------------------------
# 7. Final prediction
# ------------------------------------------------------------

nb_predictions = (
    naive_bayes_model
    .predict(X_test_nb)
)

nb_accuracy = accuracy_score(
    y_text_test,
    nb_predictions
)

nb_precision = precision_score(
    y_text_test,
    nb_predictions,
    zero_division=0
)

nb_recall = recall_score(
    y_text_test,
    nb_predictions,
    zero_division=0
)

nb_f1 = f1_score(
    y_text_test,
    nb_predictions,
    zero_division=0
)

print("\nFINAL NAIVE BAYES RESULTS")
print("=" * 60)

print(
    f"Alpha     : {best_alpha}"
)

print(
    f"Accuracy  : {nb_accuracy:.4f}"
)

print(
    f"Precision : {nb_precision:.4f}"
)

print(
    f"Recall    : {nb_recall:.4f}"
)

print(
    f"F1 Score  : {nb_f1:.4f}"
)

# ------------------------------------------------------------
# 8. Confusion matrix
# ------------------------------------------------------------

cm_nb = confusion_matrix(
    y_text_test,
    nb_predictions
)

print("\nCONFUSION MATRIX")
print("=" * 60)
print(cm_nb)

# ------------------------------------------------------------
# 9. Classification report
# ------------------------------------------------------------

print("\nCLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_text_test,
        nb_predictions,
        target_names=[
            "Negative",
            "Positive"
        ],
        zero_division=0
    )
)

# ------------------------------------------------------------
# 10. Display sample predictions
# ------------------------------------------------------------

sample_predictions = pd.DataFrame({
    "ReviewText": X_text_test.values,
    "Actual": y_text_test.values,
    "Predicted": nb_predictions
})

sample_predictions["Actual"] = (
    sample_predictions["Actual"]
    .map({
        0: "Negative",
        1: "Positive"
    })
)

sample_predictions["Predicted"] = (
    sample_predictions["Predicted"]
    .map({
        0: "Negative",
        1: "Positive"
    })
)

print("\nSAMPLE PREDICTIONS")
print("=" * 60)

display(
    sample_predictions.head(10)
)

# ------------------------------------------------------------
# 11. Save model and vectorizer
# ------------------------------------------------------------

from pathlib import Path

MODELS = Path(
    "/content/PACKVOTE/models"
)

PROCESSED = Path(
    "/content/PACKVOTE/data/processed"
)

MODELS.mkdir(
    parents=True,
    exist_ok=True
)

PROCESSED.mkdir(
    parents=True,
    exist_ok=True
)

joblib.dump(
    naive_bayes_model,
    MODELS / "naive_bayes_review_classifier.pkl"
)

joblib.dump(
    nb_vectorizer,
    MODELS / "tfidf_review_vectorizer.pkl"
)

nb_results.to_csv(
    PROCESSED / "naive_bayes_results.csv",
    index=False
)

# Save test predictions
sample_predictions.to_csv(
    PROCESSED / "naive_bayes_test_predictions.csv",
    index=False
)

print("\n✅ Naive Bayes model saved")
print(
    "Model     :",
    MODELS / "naive_bayes_review_classifier.pkl"
)

print(
    "Vectorizer:",
    MODELS / "tfidf_review_vectorizer.pkl"
)

print(
    "\n✅ STEP 10 COMPLETED — NAIVE BAYES REVIEW CLASSIFICATION"
)
```

```python
# ============================================================
# STEP 12 — PACKVOTE GROUP CONSENSUS ENGINE
# ============================================================
#
# Combines:
#   K-Means          -> Traveler profiles
#   KNN              -> Expected experience / similarity
#   Decision Tree    -> Destination suitability
#   Linear Regression-> Accommodation cost
#   Ratings/Popularity
#   Group preferences
#
# The Group Consensus Engine itself is a scoring layer,
# NOT another ML algorithm.
# ============================================================

import numpy as np
import pandas as pd
import joblib
from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ------------------------------------------------------------
# 1. Required folders
# ------------------------------------------------------------

BASE = Path("/content/PACKVOTE")
PROCESSED = BASE / "data" / "processed"
MODELS = BASE / "models"

PROCESSED.mkdir(parents=True, exist_ok=True)
MODELS.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. IMPORTANT KNN CORRECTION
# ------------------------------------------------------------
# Earlier KNN features contained ExperienceRating itself,
# which would cause target leakage.
#
# For the actual PACKVOTE recommendation engine we rebuild
# KNN without ExperienceRating as an input feature.

pref_features = interaction_features[
    "Preferences_Clean"
].apply(
    lambda text: pd.Series({
        "Pref_Beach": int("beach" in str(text).lower()),
        "Pref_Historical": int("historical" in str(text).lower()),
        "Pref_Nature": int("nature" in str(text).lower()),
        "Pref_Adventure": int("adventure" in str(text).lower()),
        "Pref_City": int("city" in str(text).lower())
    })
)

type_features = pd.get_dummies(
    interaction_features["Type"],
    prefix="Type",
    dtype=int
)

# KNN input features — NO ExperienceRating
corrected_knn_features = pd.concat(
    [
        interaction_features[
            [
                "NumberOfAdults",
                "NumberOfChildren",
                "Group_Size",
                "Popularity",
                "Preference_Match"
            ]
        ].reset_index(drop=True),

        pref_features.reset_index(drop=True),

        type_features.reset_index(drop=True)
    ],
    axis=1
)

corrected_knn_target = (
    interaction_features["ExperienceRating"]
    .astype(float)
    .reset_index(drop=True)
)

corrected_knn_features = corrected_knn_features.replace(
    [np.inf, -np.inf],
    np.nan
).fillna(0)

print("Corrected KNN feature matrix:",
      corrected_knn_features.shape)

# ------------------------------------------------------------
# 3. Train corrected KNN
# ------------------------------------------------------------

X_train_knn, X_test_knn, y_train_knn, y_test_knn = (
    train_test_split(
        corrected_knn_features,
        corrected_knn_target,
        test_size=0.20,
        random_state=42
    )
)

corrected_knn_scaler = StandardScaler()

X_train_knn_scaled = (
    corrected_knn_scaler
    .fit_transform(X_train_knn)
)

X_test_knn_scaled = (
    corrected_knn_scaler
    .transform(X_test_knn)
)

# Test a few K values
k_candidates = [3, 5, 7, 9, 11]

knn_check = []

for k in k_candidates:

    temp_model = KNeighborsRegressor(
        n_neighbors=k,
        weights="distance"
    )

    temp_model.fit(
        X_train_knn_scaled,
        y_train_knn
    )

    temp_pred = temp_model.predict(
        X_test_knn_scaled
    )

    knn_check.append({
        "K": k,
        "MAE": mean_absolute_error(
            y_test_knn,
            temp_pred
        ),
        "RMSE": mean_squared_error(
            y_test_knn,
            temp_pred
        ) ** 0.5,
        "R2": r2_score(
            y_test_knn,
            temp_pred
        )
    })

knn_check = pd.DataFrame(knn_check)

best_knn_row = knn_check.loc[
    knn_check["MAE"].idxmin()
]

best_knn_k = int(
    best_knn_row["K"]
)

# Final corrected KNN
packvote_knn = KNeighborsRegressor(
    n_neighbors=best_knn_k,
    weights="distance"
)

packvote_knn.fit(
    X_train_knn_scaled,
    y_train_knn
)

knn_check.to_csv(
    PROCESSED / "packvote_corrected_knn_results.csv",
    index=False
)

joblib.dump(
    packvote_knn,
    MODELS / "packvote_knn_model.pkl"
)

joblib.dump(
    corrected_knn_scaler,
    MODELS / "packvote_knn_scaler.pkl"
)

joblib.dump(
    corrected_knn_features.columns.tolist(),
    MODELS / "packvote_knn_features.pkl"
)

print("\n✅ Leakage-corrected KNN ready")
print("Selected K:", best_knn_k)
print("MAE:", round(best_knn_row["MAE"], 4))
print("RMSE:", round(best_knn_row["RMSE"], 4))
print("R²:", round(best_knn_row["R2"], 4))

# ------------------------------------------------------------
# 4. Build unique destination candidates
# ------------------------------------------------------------

destination_candidates = (
    destinations_p
    .groupby(
        [
            "Name",
            "State",
            "Type",
            "BestTimeToVisit"
        ],
        as_index=False
    )
    .agg(
        DestinationID=("DestinationID", "first"),
        Popularity=("Popularity", "mean")
    )
)

# Add rating information
destination_rating = (
    reviews_ml
    .merge(
        destinations_p[
            ["DestinationID", "Name"]
        ],
        on="DestinationID",
        how="left"
    )
    .groupby("Name")
    .agg(
        Avg_Rating=("Rating", "mean"),
        Review_Count=("ReviewID", "count")
    )
    .reset_index()
)

destination_candidates = (
    destination_candidates
    .merge(
        destination_rating,
        on="Name",
        how="left"
    )
)

destination_candidates[
    ["Avg_Rating", "Review_Count"]
] = destination_candidates[
    ["Avg_Rating", "Review_Count"]
].fillna(0)

# Popularity normalization
pop_min = destination_candidates["Popularity"].min()
pop_max = destination_candidates["Popularity"].max()

destination_candidates["Popularity_Norm"] = (
    destination_candidates["Popularity"] - pop_min
) / (
    pop_max - pop_min + 1e-9
)

destination_candidates["Rating_Norm"] = (
    destination_candidates["Avg_Rating"] / 5
)

# ------------------------------------------------------------
# 5. Planning-city bridge
# ------------------------------------------------------------
# Dataset 2 and Dataset 3 do NOT share Dataset 1 destination IDs.
#
# These mappings are used only as a planning bridge for places
# whose destination name clearly corresponds to a city.
#
# If no safe city bridge exists, budget_fit becomes neutral
# instead of inventing a cost.

planning_city_map = {
    "Goa Beaches": "Goa",
    "Jaipur City": "Jaipur",
    "Taj Mahal": "Agra",
    "Leh Ladakh": "Leh Ladakh"
}

destination_candidates["Planning_City"] = (
    destination_candidates["Name"]
    .map(planning_city_map)
)

# ------------------------------------------------------------
# 6. Prepare accommodation information
# ------------------------------------------------------------

city_cost_summary = (
    travel_cost_p[
        travel_cost_p["Cost_Average"].notna()
    ]
    .groupby("City", as_index=False)
    .agg(
        Cost_Min=("Cost_Min", "min"),
        Cost_Max=("Cost_Max", "max"),
        Cost_Average=("Cost_Average", "mean")
    )
)

destination_candidates = (
    destination_candidates
    .merge(
        city_cost_summary,
        left_on="Planning_City",
        right_on="City",
        how="left"
    )
    .drop(columns=["City"])
)

# ------------------------------------------------------------
# 7. K-Means cluster profile
# ------------------------------------------------------------

cluster_preference_columns = [
    "Pref_Beach",
    "Pref_Historical",
    "Pref_Nature",
    "Pref_Adventure",
    "Pref_City"
]

cluster_preference_profile = (
    traveler_features
    .groupby("Traveler_Cluster")[
        cluster_preference_columns
    ]
    .mean()
)

type_to_preference = {
    "Beach": "Pref_Beach",
    "Historical": "Pref_Historical",
    "Nature": "Pref_Nature",
    "Adventure": "Pref_Adventure",
    "City": "Pref_City"
}

# ------------------------------------------------------------
# 8. Build KNN input vector
# ------------------------------------------------------------

def create_knn_input(
    preference_text,
    destination_type,
    popularity,
    adults=1,
    children=0
):

    text = str(preference_text).lower()

    preference_match = int(
        str(destination_type).lower()
        in text
    )

    row = {
        "NumberOfAdults": adults,
        "NumberOfChildren": children,
        "Group_Size": adults + children,
        "Popularity": popularity,
        "Preference_Match": preference_match,

        "Pref_Beach":
            int("beach" in text),

        "Pref_Historical":
            int("historical" in text),

        "Pref_Nature":
            int("nature" in text),

        "Pref_Adventure":
            int("adventure" in text),

        "Pref_City":
            int("city" in text)
    }

    # Add all type columns required by training
    for col in corrected_knn_features.columns:

        if col.startswith("Type_"):

            row[col] = int(
                col.replace("Type_", "").lower()
                ==
                str(destination_type).lower()
            )

    # Guarantee exact training-column order
    result = pd.DataFrame([row])

    result = result.reindex(
        columns=corrected_knn_features.columns,
        fill_value=0
    )

    return result.astype(float)

# ------------------------------------------------------------
# 9. Build Decision Tree input vector
# ------------------------------------------------------------

# Use the same columns that were used while training
dt_feature_columns = X_dt.columns.tolist()

def create_tree_input(
    preference_text,
    destination_type,
    popularity,
    adults=1,
    children=0
):

    text = str(preference_text).lower()

    row = {
        "NumberOfAdults": adults,
        "NumberOfChildren": children,
        "Group_Size": adults + children,
        "Popularity": popularity,
        "Preference_Match": int(
            str(destination_type).lower()
            in text
        ),

        "Pref_Beach": int("beach" in text),
        "Pref_Historical": int("historical" in text),
        "Pref_Nature": int("nature" in text),
        "Pref_Adventure": int("adventure" in text),
        "Pref_City": int("city" in text)
    }

    for col in dt_feature_columns:

        if col.startswith("Type_"):

            row[col] = int(
                col.replace("Type_", "").lower()
                ==
                str(destination_type).lower()
            )

    result = pd.DataFrame([row])

    result = result.reindex(
        columns=dt_feature_columns,
        fill_value=0
    )

    return result.astype(float)

# ------------------------------------------------------------
# 10. Group Consensus Engine
# ------------------------------------------------------------

def recommend_group(
    group_preferences,
    budget_per_person,
    accommodation_type="Hotel",
    adults_per_traveler=1,
    children_per_traveler=0
):
    """
    group_preferences:
        Example:
        [
            "Beach, Historical",
            "Nature, Adventure",
            "City, Historical",
            "Nature, Adventure"
        ]

    budget_per_person:
        Accommodation budget reference in INR.
    """

    if len(group_preferences) == 0:
        raise ValueError(
            "At least one traveler preference is required."
        )

    results = []

    for _, destination in (
        destination_candidates.iterrows()
    ):

        individual_scores = []
        predicted_experiences = []
        suitability_scores = []
        preference_scores = []
        cluster_scores = []

        # --------------------------------------------
        # Score destination for every traveler
        # --------------------------------------------

        for preference in group_preferences:

            # KNN
            knn_input = create_knn_input(
                preference_text=preference,
                destination_type=destination["Type"],
                popularity=destination["Popularity"],
                adults=adults_per_traveler,
                children=children_per_traveler
            )

            knn_input_scaled = (
                corrected_knn_scaler
                .transform(knn_input)
            )

            predicted_experience = float(
                packvote_knn.predict(
                    knn_input_scaled
                )[0]
            )

            predicted_experience = np.clip(
                predicted_experience,
                1,
                5
            )

            experience_score = (
                predicted_experience - 1
            ) / 4

            # Decision Tree
            tree_input = create_tree_input(
                preference_text=preference,
                destination_type=destination["Type"],
                popularity=destination["Popularity"],
                adults=adults_per_traveler,
                children=children_per_traveler
            )

            suitability_probability = float(
                decision_tree_model
                .predict_proba(tree_input)[0, 1]
            )

            # Preference match
            preference_match = int(
                str(destination["Type"]).lower()
                in str(preference).lower()
            )

            # K-Means cluster alignment
            temp_profile = pd.DataFrame([{
                col:
                    int(
                        col.replace("Pref_", "").lower()
                        in str(preference).lower()
                    )
                for col in cluster_preference_columns
            }])

            # Find closest traveler cluster based on preferences
            cluster_distances = (
                (
                    cluster_preference_profile
                    - temp_profile.iloc[0]
                )
                ** 2
            ).sum(axis=1)

            nearest_cluster = (
                cluster_distances.idxmin()
            )

            pref_col = type_to_preference.get(
                destination["Type"],
                None
            )

            if pref_col is not None:
                cluster_support = float(
                    cluster_preference_profile
                    .loc[
                        nearest_cluster,
                        pref_col
                    ]
                )
            else:
                cluster_support = 0.5

            # Individual score
            rating_score = float(
                destination["Rating_Norm"]
            )

            popularity_score = float(
                destination["Popularity_Norm"]
            )

            individual_score = 100 * (
                0.30 * preference_match +
                0.20 * experience_score +
                0.20 * suitability_probability +
                0.10 * rating_score +
                0.10 * popularity_score +
                0.10 * cluster_support
            )

            individual_scores.append(
                individual_score
            )

            predicted_experiences.append(
                predicted_experience
            )

            suitability_scores.append(
                suitability_probability
            )

            preference_scores.append(
                preference_match
            )

            cluster_scores.append(
                cluster_support
            )

        # --------------------------------------------
        # Group-level scores
        # --------------------------------------------

        average_group_score = np.mean(
            individual_scores
        )

        least_satisfied_member = np.min(
            individual_scores
        )

        group_preference_score = np.mean(
            preference_scores
        )

        cluster_consensus = np.mean(
            cluster_scores
        )

        # --------------------------------------------
        # Budget score
        # --------------------------------------------

        predicted_cost = np.nan
        budget_fit = 0.5
        budget_status = "Cost data unavailable"

        planning_city = destination[
            "Planning_City"
        ]

        if pd.notna(planning_city):

            try:

                cost_input = pd.DataFrame({
                    "City": [planning_city],
                    "Accomadation_Type": [
                        accommodation_type
                    ]
                })

                predicted_cost = float(
                    linear_regression_model
                    .predict(cost_input)[0]
                )

                predicted_cost = max(
                    0,
                    predicted_cost
                )

                if predicted_cost <= budget_per_person:

                    budget_fit = 1.0
                    budget_status = "Within budget"

                else:

                    budget_fit = (
                        budget_per_person /
                        (predicted_cost + 1e-9)
                    )

                    budget_fit = min(
                        max(budget_fit, 0),
                        1
                    )

                    budget_status = "Above budget"

            except Exception:

                budget_status = (
                    "Cost prediction unavailable"
                )

        # --------------------------------------------
        # FINAL GROUP CONSENSUS SCORE
        # --------------------------------------------
        #
        # 70% average satisfaction
        # 15% least-satisfied traveler
        # 15% budget fit
        #

        final_group_score = (
            0.70 * average_group_score +
            0.15 * least_satisfied_member +
            0.15 * (budget_fit * 100)
        )

        results.append({

            "Destination":
                destination["Name"],

            "State":
                destination["State"],

            "Type":
                destination["Type"],

            "Group_Compatibility":
                round(
                    final_group_score,
                    2
                ),

            "Average_Member_Score":
                round(
                    average_group_score,
                    2
                ),

            "Minimum_Member_Score":
                round(
                    least_satisfied_member,
                    2
                ),

            "Group_Preference_Match":
                round(
                    group_preference_score * 100,
                    2
                ),

            "KMeans_Cluster_Support":
                round(
                    cluster_consensus * 100,
                    2
                ),

            "Predicted_Experience":
                round(
                    np.mean(
                        predicted_experiences
                    ),
                    2
                ),

            "Suitability":
                round(
                    np.mean(
                        suitability_scores
                    ) * 100,
                    2
                ),

            "Popularity":
                round(
                    destination["Popularity"],
                    2
                ),

            "Average_Rating":
                round(
                    destination["Avg_Rating"],
                    2
                ),

            "Best_Time":
                destination["BestTimeToVisit"],

            "Planning_City":
                planning_city,

            "Predicted_Accommodation_Cost":
                round(
                    predicted_cost,
                    2
                )
                if not pd.isna(predicted_cost)
                else np.nan,

            "Budget_Fit":
                round(
                    budget_fit * 100,
                    2
                ),

            "Budget_Status":
                budget_status,

            "Accommodation_Min":
                destination["Cost_Min"],

            "Accommodation_Max":
                destination["Cost_Max"]
        })

    result_df = pd.DataFrame(
        results
    )

    result_df = (
        result_df
        .sort_values(
            "Group_Compatibility",
            ascending=False
        )
        .reset_index(drop=True)
    )

    result_df["Rank"] = (
        result_df.index + 1
    )

    # Put Rank first
    result_df = result_df[
        ["Rank"] +
        [
            col for col in result_df.columns
            if col != "Rank"
        ]
    ]

    return result_df

# ------------------------------------------------------------
# 11. Test PACKVOTE with a sample group
# ------------------------------------------------------------

sample_group = [
    "Beach, Historical",
    "Nature, Adventure",
    "City, Historical",
    "Nature, Adventure"
]

sample_budget = 5000

recommendations = recommend_group(
    group_preferences=sample_group,
    budget_per_person=sample_budget,
    accommodation_type="Hotel"
)

# ------------------------------------------------------------
# 12. Display final ranking
# ------------------------------------------------------------

print("\n========================================================")
print("             PACKVOTE GROUP RECOMMENDATIONS")
print("========================================================")

display(
    recommendations[
        [
            "Rank",
            "Destination",
            "State",
            "Type",
            "Group_Compatibility",
            "Group_Preference_Match",
            "Predicted_Experience",
            "Suitability",
            "Average_Rating",
            "Popularity",
            "Budget_Status",
            "Predicted_Accommodation_Cost",
            "Best_Time"
        ]
    ].head(10)
)

# ------------------------------------------------------------
# 13. Save recommendations
# ------------------------------------------------------------

recommendations.to_csv(
    PROCESSED / "packvote_group_recommendations.csv",
    index=False
)

# Save the group input used for this run
group_input = pd.DataFrame({
    "Traveler": [
        f"Traveler {i+1}"
        for i in range(
            len(sample_group)
        )
    ],
    "Preference": sample_group
})

group_input.to_csv(
    PROCESSED / "sample_group_input.csv",
    index=False
)

# ------------------------------------------------------------
# 14. Final top destination explanation
# ------------------------------------------------------------

top = recommendations.iloc[0]

print("\n========================================================")
print("                 FINAL RECOMMENDATION")
print("========================================================")

print(
    f"Rank              : {int(top['Rank'])}"
)

print(
    f"Destination       : {top['Destination']}"
)

print(
    f"State             : {top['State']}"
)

print(
    f"Type              : {top['Type']}"
)

print(
    f"Group Compatibility: "
    f"{top['Group_Compatibility']:.2f}/100"
)

print(
    f"Group Preference Match: "
    f"{top['Group_Preference_Match']:.2f}%"
)

print(
    f"Predicted Experience: "
    f"{top['Predicted_Experience']:.2f}/5"
)

print(
    f"Suitability        : "
    f"{top['Suitability']:.2f}%"
)

print(
    f"Average Rating     : "
    f"{top['Average_Rating']:.2f}/5"
)

print(
    f"Best Time          : "
    f"{top['Best_Time']}"
)

print(
    f"Budget Status      : "
    f"{top['Budget_Status']}"
)

if not pd.isna(
    top["Predicted_Accommodation_Cost"]
):

    print(
        f"Predicted Accommodation Cost: "
        f"₹{top['Predicted_Accommodation_Cost']:,.2f}"
    )

print(
    f"K-Means Cluster Support: "
    f"{top['KMeans_Cluster_Support']:.2f}%"
)

print("\n✅ STEP 12 COMPLETED — GROUP CONSENSUS ENGINE")
```

```python
# ============================================================
# STEP 14 — TOURIST SPOT INTEGRATION
# ============================================================

import pandas as pd
import numpy as np
import re
from pathlib import Path

BASE = Path("/content/PACKVOTE")
PROCESSED = BASE / "data" / "processed"
CITYWISE_DIR = BASE / "datasets" / "3rd" / "Data" / "Citywise Destinations"

PROCESSED.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Load Step 13 final ranking
# ------------------------------------------------------------

ranking_file = PROCESSED / "final_destination_ranking.csv"

if not ranking_file.exists():
    raise FileNotFoundError(
        "final_destination_ranking.csv not found. "
        "Please complete Step 13 first."
    )

final_ranking = pd.read_csv(ranking_file)

print("✅ Step 13 ranking loaded")
print("Number of ranked destinations:", len(final_ranking))

# ------------------------------------------------------------
# 2. Check citywise dataset folder
# ------------------------------------------------------------

if not CITYWISE_DIR.exists():
    raise FileNotFoundError(
        f"Citywise folder not found:\n{CITYWISE_DIR}"
    )

city_files = sorted(CITYWISE_DIR.glob("*.csv"))

print("\nCitywise files found:", len(city_files))

for f in city_files:
    print(" -", f.name)

# ------------------------------------------------------------
# 3. Helper: normalize city names
# ------------------------------------------------------------

def normalize_city(city):

    if pd.isna(city):
        return ""

    city = str(city).strip().lower()

    city = re.sub(
        r"destinations[_\-\s]*",
        "",
        city
    )

    city = re.sub(
        r"[^a-z0-9\s]",
        "",
        city
    )

    city = re.sub(
        r"\s+",
        " ",
        city
    ).strip()

    aliases = {
        "bengaluru": "bangalore",
        "banglore": "bangalore",
        "calcutta": "kolkata",
        "bombay": "mumbai"
    }

    return aliases.get(
        city,
        city
    )

# ------------------------------------------------------------
# 4. Load all citywise attraction files
# ------------------------------------------------------------

all_attractions = []

for file in city_files:

    try:
        df = pd.read_csv(file)
    except Exception as e:
        print(
            f"⚠️ Could not read {file.name}: {e}"
        )
        continue

    # Find likely columns
    cols_lower = {
        str(c).strip().lower(): c
        for c in df.columns
    }

    name_col = None
    char_col = None
    lat_col = None
    lon_col = None

    for candidate in [
        "name",
        "poi_name",
        "poi name",
        "destination",
        "attraction",
        "tourist spot",
        "touristspot",
        "place"
    ]:
        if candidate in cols_lower:
            name_col = cols_lower[candidate]
            break

    for candidate in [
        "characteristics",
        "characteristic",
        "features",
        "theme"
    ]:
        if candidate in cols_lower:
            char_col = cols_lower[candidate]
            break

    for candidate in [
        "latitude",
        "lat"
    ]:
        if candidate in cols_lower:
            lat_col = cols_lower[candidate]
            break

    for candidate in [
        "longitude",
        "long",
        "lon"
    ]:
        if candidate in cols_lower:
            lon_col = cols_lower[candidate]
            break

    # Fallback to first object column for name
    if name_col is None:

        object_cols = (
            df.select_dtypes(
                include="object"
            ).columns.tolist()
        )

        if object_cols:
            name_col = object_cols[0]

    if name_col is None:
        print(
            f"⚠️ Skipping {file.name}: "
            "no attraction-name column found."
        )
        continue

    # City is derived from filename because these are citywise files
    city_name = file.stem

    # Remove "Destinations_" prefix if present
    city_name = re.sub(
        r"^Destinations[_\-]?",
        "",
        city_name,
        flags=re.IGNORECASE
    )

    for _, row in df.iterrows():

        attraction_name = str(
            row.get(
                name_col,
                "Unknown"
            )
        ).strip()

        characteristics = (
            str(
                row.get(
                    char_col,
                    ""
                )
            ).strip()
            if char_col is not None
            else ""
        )

        latitude = (
            row.get(
                lat_col,
                np.nan
            )
            if lat_col is not None
            else np.nan
        )

        longitude = (
            row.get(
                lon_col,
                np.nan
            )
            if lon_col is not None
            else np.nan
        )

        all_attractions.append({
            "City": city_name,
            "City_Normalized": normalize_city(city_name),
            "Attraction": attraction_name,
            "Characteristics": characteristics,
            "Latitude": latitude,
            "Longitude": longitude,
            "Source_File": file.name
        })

# ------------------------------------------------------------
# 5. Create unified attraction table
# ------------------------------------------------------------

citywise_attractions = pd.DataFrame(
    all_attractions
)

citywise_attractions = (
    citywise_attractions
    .drop_duplicates(
        subset=[
            "City",
            "Attraction"
        ]
    )
    .reset_index(drop=True)
)

print("\n✅ Unified attraction dataset:")
print(
    "Rows:",
    len(citywise_attractions)
)

print(
    "Cities:",
    citywise_attractions["City"].nunique()
)

display(
    citywise_attractions.head(10)
)

# ------------------------------------------------------------
# 6. Define safe planning-city bridges
# ------------------------------------------------------------
#
# These are ONLY used when the destination name clearly maps
# to a city represented in Dataset 3.
#
# We do NOT invent a city for destinations that have no
# defensible city-level mapping.

safe_city_map = {
    "Goa Beaches": "Goa",
    "Jaipur City": "Jaipur",
    "Taj Mahal": "Agra",
    "Leh Ladakh": "Leh Ladakh"
}

# Normalize the planning-city bridge
safe_city_map_normalized = {
    key: normalize_city(value)
    for key, value in safe_city_map.items()
}

# ------------------------------------------------------------
# 7. Match ranked destinations to Dataset 3 cities
# ------------------------------------------------------------

final_destination_output = []

for _, row in final_ranking.iterrows():

    destination = row["Destination"]

    matched_city = safe_city_map_normalized.get(
        destination,
        ""
    )

    matched_spots = pd.DataFrame()

    if matched_city:

        matched_spots = (
            citywise_attractions[
                citywise_attractions[
                    "City_Normalized"
                ] == matched_city
            ]
            .copy()
        )

    # Top 8 tourist spots
    if not matched_spots.empty:

        matched_spots = (
            matched_spots
            .drop_duplicates(
                subset=["Attraction"]
            )
            .head(8)
        )

        spot_names = (
            matched_spots["Attraction"]
            .tolist()
        )

        spot_text = " | ".join(
            spot_names
        )

        spot_count = len(
            matched_spots
        )

        tourist_status = (
            "Dataset 3 city match found"
        )

    else:

        spot_text = (
            "No direct city-level match "
            "available in Dataset 3"
        )

        spot_count = 0

        tourist_status = (
            "No Dataset 3 city match"
        )

    final_destination_output.append({

        "Rank":
            row["Rank"],

        "Destination":
            destination,

        "State":
            row["State"],

        "Type":
            row["Type"],

        "Group_Compatibility":
            row["Group_Compatibility"],

        "Group_Preference_Match":
            row["Group_Preference_Match"],

        "KMeans_Cluster_Support":
            row["KMeans_Cluster_Support"],

        "Predicted_Experience":
            row["Predicted_Experience"],

        "Suitability":
            row["Suitability"],

        "Average_Rating":
            row["Average_Rating"],

        "Popularity":
            row["Popularity"],

        "Budget_Fit":
            row["Budget_Fit"],

        "Budget_Status":
            row["Budget_Status"],

        "Predicted_Accommodation_Cost":
            row["Predicted_Accommodation_Cost"],

        "Best_Time":
            row["Best_Time"],

        "Planning_City":
            matched_city
            if matched_city
            else "Not mapped",

        "Tourist_Spot_Count":
            spot_count,

        "Tourist_Spots":
            spot_text,

        "Tourist_Data_Status":
            tourist_status,

        "Why_This_Destination":
            row["Why_This_Destination"]
    })

# ------------------------------------------------------------
# 8. Final integrated output
# ------------------------------------------------------------

final_destination_output = pd.DataFrame(
    final_destination_output
)

print()
print("=" * 100)
print("           PACKVOTE — FINAL DESTINATION + TOURIST OUTPUT")
print("=" * 100)

display(
    final_destination_output.head(5)
)

# ------------------------------------------------------------
# 9. Show TOP recommendation in detail
# ------------------------------------------------------------

top = final_destination_output.iloc[0]

print()
print("=" * 100)
print("                  PACKVOTE FINAL RECOMMENDATION")
print("=" * 100)

print(
    f"Rank                  : {int(top['Rank'])}"
)

print(
    f"Destination            : {top['Destination']}"
)

print(
    f"State                  : {top['State']}"
)

print(
    f"Type                   : {top['Type']}"
)

print(
    f"Group Compatibility    : "
    f"{top['Group_Compatibility']:.2f}/100"
)

print(
    f"Group Preference Match : "
    f"{top['Group_Preference_Match']:.2f}%"
)

print(
    f"Suitability            : "
    f"{top['Suitability']:.2f}%"
)

print(
    f"Average Rating         : "
    f"{top['Average_Rating']:.2f}/5"
)

print(
    f"Budget Status          : "
    f"{top['Budget_Status']}"
)

print(
    f"Best Time              : "
    f"{top['Best_Time']}"
)

print(
    f"Planning City          : "
    f"{top['Planning_City']}"
)

print(
    f"Tourist Spots Found    : "
    f"{int(top['Tourist_Spot_Count'])}"
)

print()
print("WHY THIS DESTINATION?")
print("-" * 100)
print(
    top["Why_This_Destination"]
)

print()
print("TOURIST SPOTS")
print("-" * 100)

if top["Tourist_Spot_Count"] > 0:

    for i, spot in enumerate(
        top["Tourist_Spots"].split(" | "),
        start=1
    ):
        print(
            f"{i}. {spot}"
        )

else:

    print(
        "No direct city-level tourist-spot "
        "match is available in Dataset 3."
    )

# ------------------------------------------------------------
# 10. Save final integrated output
# ------------------------------------------------------------

final_destination_output.to_csv(
    PROCESSED /
    "packvote_final_destination_output.csv",
    index=False
)

citywise_attractions.to_csv(
    PROCESSED /
    "packvote_citywise_attractions.csv",
    index=False
)

print()
print("✅ Saved:")
print(
    PROCESSED /
    "packvote_final_destination_output.csv"
)

print(
    PROCESSED /
    "packvote_citywise_attractions.csv"
)

print()
print(
    "✅ STEP 14 COMPLETED — TOURIST SPOT INTEGRATION"
)
```

```python
import gradio as gr
import pandas as pd
import numpy as np
from pathlib import Path
import html
import os

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

BASE = Path("/content/PACKVOTE")
PROCESSED = BASE / "data" / "processed"

PROCESSED.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# 2. Make sure Step 12 recommendation function exists
# ------------------------------------------------------------

if "recommend_group" not in globals():

    raise RuntimeError(
        "The recommend_group() function is not available.\n\n"
        "Please run Step 12 first, then run Step 15."
    )

# ------------------------------------------------------------
# 3. Safe city mapping for Dataset 3
# ------------------------------------------------------------

safe_city_map = {
    "Goa Beaches": "goa",
    "Jaipur City": "jaipur",
    "Taj Mahal": "agra",
    "Leh Ladakh": "leh ladakh"
}

# ------------------------------------------------------------
# 4. Load tourist attraction table if available
# ------------------------------------------------------------

attraction_file = (
    PROCESSED /
    "packvote_citywise_attractions.csv"
)

if attraction_file.exists():

    citywise_attractions = pd.read_csv(
        attraction_file
    )

else:

    citywise_attractions = pd.DataFrame(
        columns=[
            "City",
            "City_Normalized",
            "Attraction",
            "Characteristics"
        ]
    )


def normalize_city(city):

    if pd.isna(city):
        return ""

    city = str(city).strip().lower()

    aliases = {
        "bengaluru": "bangalore",
        "banglore": "bangalore",
        "calcutta": "kolkata",
        "bombay": "mumbai"
    }

    return aliases.get(
        city,
        city
    )


# ------------------------------------------------------------
# 5. Generate recommendation cards
# ------------------------------------------------------------

def make_destination_card(
    rank,
    row
):

    destination = html.escape(
        str(row["Destination"])
    )

    state = html.escape(
        str(row["State"])
    )

    dtype = html.escape(
        str(row["Type"])
    )

    score = float(
        row["Group_Compatibility"]
    )

    experience = float(
        row["Predicted_Experience"]
    )

    suitability = float(
        row["Suitability"]
    )

    rating = float(
        row["Average_Rating"]
    )

    best_time = html.escape(
        str(row["Best_Time"])
    )

    budget_status = html.escape(
        str(row["Budget_Status"])
    )

    explanation = html.escape(
        str(row["Why_This_Destination"])
    )

    cost = row.get(
        "Predicted_Accommodation_Cost",
        np.nan
    )

    if pd.notna(cost):
        cost_text = (
            f"₹{float(cost):,.0f}"
        )
    else:
        cost_text = "N/A"

    return f"""
    <div class="destination-card">

        <div class="card-top">

            <div class="rank-circle">
                #{rank}
            </div>

            <div class="destination-title">
                <h2>{destination}</h2>
                <div class="location">
                    📍 {state} &nbsp; • &nbsp; {dtype}
                </div>
            </div>

            <div class="score">
                {score:.1f}
                <span>/100</span>
            </div>

        </div>

        <div class="metrics">

            <div class="metric">
                <span>Group Match</span>
                <strong>
                    {float(row["Group_Preference_Match"]):.1f}%
                </strong>
            </div>

            <div class="metric">
                <span>Suitability</span>
                <strong>
                    {suitability:.1f}%
                </strong>
            </div>

            <div class="metric">
                <span>Experience</span>
                <strong>
                    {experience:.2f}/5
                </strong>
            </div>

            <div class="metric">
                <span>Rating</span>
                <strong>
                    ⭐ {rating:.2f}
                </strong>
            </div>

        </div>

        <div class="info-row">

            <div>
                🏨
                <b>Predicted accommodation:</b>
                {cost_text}
            </div>

            <div>
                💰
                <b>Budget:</b>
                {budget_status}
            </div>

            <div>
                🌤️
                <b>Best time:</b>
                {best_time}
            </div>

        </div>

        <div class="why-box">
            <b>Why this destination?</b><br>
            {explanation}
        </div>

    </div>
    """


# ------------------------------------------------------------
# 6. Generate tourist spot section
# ------------------------------------------------------------

def get_tourist_spots(
    destination
):

    city = safe_city_map.get(
        str(destination),
        ""
    )

    if not city:
        return (
            "<div class='no-data'>"
            "Tourist-spot city mapping is not available "
            "for this destination in the selected datasets."
            "</div>"
        )

    if citywise_attractions.empty:
        return (
            "<div class='no-data'>"
            "Tourist-spot data is not loaded."
            "</div>"
        )

    if "City_Normalized" in citywise_attractions.columns:

        matches = citywise_attractions[
            citywise_attractions[
                "City_Normalized"
            ]
            .astype(str)
            .str.lower()
            == city
        ].copy()

    else:

        matches = citywise_attractions[
            citywise_attractions[
                "City"
            ]
            .astype(str)
            .str.lower()
            == city
        ].copy()

    matches = matches.head(8)

    if matches.empty:

        return (
            "<div class='no-data'>"
            "No matching tourist spots were found "
            "in Dataset 3."
            "</div>"
        )

    cards = ""

    for i, (_, spot) in enumerate(
        matches.iterrows(),
        start=1
    ):

        attraction = html.escape(
            str(
                spot.get(
                    "Attraction",
                    "Unknown"
                )
            )
        )

        characteristics = html.escape(
            str(
                spot.get(
                    "Characteristics",
                    ""
                )
            )
        )

        cards += f"""
        <div class="spot-card">
            <div class="spot-number">
                {i}
            </div>

            <div>
                <b>{attraction}</b>
                <div class="spot-characteristics">
                    {characteristics}
                </div>
            </div>
        </div>
        """

    return f"""
    <div class="spots-container">
        {cards}
    </div>
    """


# ------------------------------------------------------------
# 7. Main recommendation function
# ------------------------------------------------------------

def run_packvote(
    traveler_count,
    traveler1,
    traveler2,
    traveler3,
    traveler4,
    budget,
    accommodation_type
):

    traveler_count = int(
        traveler_count
    )

    all_preferences = [
        traveler1,
        traveler2,
        traveler3,
        traveler4
    ]

    # Use only requested number of travelers
    preferences = [
        str(x).strip()
        for x in all_preferences[
            :traveler_count
        ]
        if str(x).strip()
    ]

    if len(preferences) == 0:

        return (
            "<div class='error-box'>"
            "Please enter at least one traveler preference."
            "</div>",
            pd.DataFrame(),
            "<div class='error-box'>"
            "No tourist information available."
            "</div>",
            None
        )

    # --------------------------------------------------------
    # Run PACKVOTE
    # --------------------------------------------------------

    result = recommend_group(
        group_preferences=preferences,
        budget_per_person=float(budget),
        accommodation_type=accommodation_type
    )

    result = result.copy()

    # Add the 'Why_This_Destination' column
    # Ensure create_reason is defined and available in the global scope (e.g., by running Step 13 first)
    if 'Why_This_Destination' not in result.columns:
        try:
            result["Why_This_Destination"] = result.apply(create_reason, axis=1)
        except NameError:
            # Fallback if create_reason is not defined, should not happen if Step 13 ran.
            result["Why_This_Destination"] = "Reasoning not available. Please ensure all previous steps are run."

    # Make sure rank exists
    result["Rank"] = (
        np.arange(
            1,
            len(result) + 1
        )
    )

    # --------------------------------------------------------
    # Save live recommendation
    # --------------------------------------------------------

    live_file = (
        PROCESSED /
        "live_packvote_recommendations.csv"
    )

    result.to_csv(
        live_file,
        index=False
    )

    # --------------------------------------------------------
    # Build recommendation cards
    # --------------------------------------------------------

    cards = ""

    top_results = result.head(3)

    for _, row in top_results.iterrows():

        cards += make_destination_card(
            int(row["Rank"]),
            row
        )

    recommendations_html = f"""
    <div class="result-header">
        <h1>🎯 Top PACKVOTE Recommendations</h1>
        <p>
            Recommendations generated for
            <b>{len(preferences)} traveler(s)</b>
            using the PACKVOTE ML pipeline.
        </p>
    </div>

    {cards}
    """

    # --------------------------------------------------------
    # Recommendation table
    # --------------------------------------------------------

    table_columns = [
        "Rank",
        "Destination",
        "State",
        "Type",
        "Group_Compatibility",
        "Predicted_Experience",
        "Suitability",
        "Average_Rating",
        "Budget_Status",
        "Best_Time"
    ]

    table_columns = [
        col
        for col in table_columns
        if col in result.columns
    ]

    result_table = result[
        table_columns
    ].head(10)

    # --------------------------------------------------------
    # Tourist spots for top destination
    # --------------------------------------------------------

    top_destination = result.iloc[0]["Destination"]

    tourist_html = f"""
    <div class="result-header">
        <h1>📍 Tourist Spots — {html.escape(str(top_destination))}</h1>
        <p>
            Tourist information is taken from Dataset 3
            through the supported city-level mapping.
        </p>
    </div>

    {get_tourist_spots(top_destination)}
    """

    return (
        recommendations_html,
        result_table,
        tourist_html,
        str(live_file)
    )


# ------------------------------------------------------------
# 8. Custom CSS
# ------------------------------------------------------------

css = """
body {
    background: #f5f8fc;
}

.gradio-container {
    max-width: 1400px !important;
    margin: auto;
}

.app-title {
    text-align: center;
    padding: 25px;
    border-radius: 18px;
    background: linear-gradient(
        135deg,
        #0b3d91,
        #1769d1
    );
    color: white;
    margin-bottom: 18px;
}

.app-title h1 {
    font-size: 38px;
    margin-bottom: 5px;
}

.app-title p {
    font-size: 17px;
    opacity: 0.95;
}

.section-title {
    font-size: 24px;
    font-weight: 700;
    color: #12345b;
}

.destination-card {
    background: white;
    border-radius: 18px;
    padding: 22px;
    margin: 18px 0;
    box-shadow: 0 6px 20px rgba(
        30, 70, 120, 0.10
    );
    border: 1px solid #dce7f3;
}

.card-top {
    display: flex;
    align-items: center;
    gap: 15px;
}

.rank-circle {
    width: 44px;
    height: 44px;
    border-radius: 50%;
    background: #1769d1;
    color: white;
    display: flex;
    align-items: center;
    justify-content: center;
    font-weight: 700;
}

.destination-title {
    flex: 1;
}

.destination-title h2 {
    margin: 0;
    color: #12345b;
}

.location {
    margin-top: 5px;
    color: #66788a;
}

.score {
    background: #e8f7ec;
    color: #14833b;
    border-radius: 12px;
    padding: 10px 15px;
    font-size: 24px;
    font-weight: 800;
}

.score span {
    font-size: 13px;
    font-weight: 500;
}

.metrics {
    display: grid;
    grid-template-columns:
        repeat(4, 1fr);
    gap: 10px;
    margin-top: 18px;
}

.metric {
    background: #f4f8fd;
    padding: 12px;
    border-radius: 12px;
}

.metric span {
    display: block;
    color: #708090;
    font-size: 12px;
}

.metric strong {
    color: #173f67;
    font-size: 18px;
}

.info-row {
    display: flex;
    flex-wrap: wrap;
    gap: 20px;
    padding: 15px 0;
    color: #40566f;
}

.why-box {
    background: #eef6ff;
    padding: 15px;
    border-radius: 12px;
    color: #274965;
}

.result-header h1 {
    color: #12345b;
}

.result-header p {
    color: #66788a;
}

.spot-card {
    display: flex;
    align-items: flex-start;
    gap: 15px;
    background: white;
    border: 1px solid #dce7f3;
    border-radius: 14px;
    padding: 14px;
    margin: 9px 0;
    box-shadow: 0 4px 12px rgba(
        30, 70, 120, 0.07
    );
    position: relative;
}

.spot-number {
    background: #1769d1;
    color: white;
    min-width: 30px;
    height: 30px;
    border-radius: 50%;
    display: flex;
    justify-content: center;
    align-items: center;
    font-weight: 700;
    flex-shrink: 0;
    margin-top: 2px; /* Adjust for better alignment */
}

.spot-characteristics {
    color: #718096;
    margin-top: 5px;
    font-size: 13px;
}

.no-data,
.error-box {
    background: white;
    border-radius: 14px;
    padding: 20px;
    color: #7a4e00;
    border: 1px solid #f1d79c;
}
"""

# ------------------------------------------------------------
# 9. Build Gradio application
# ------------------------------------------------------------

with gr.Blocks(
    title="PACKVOTE — Group Travel Recommendation"
) as demo:

    gr.HTML(
        """
        <div class="app-title">
            <h1>🏔️ PACKVOTE</h1>
            <p>
                Group Travel Recommendation & Budget-Aware
                Destination Planning System
            </p>
            <p>
                Powered by Machine Learning
            </p>
        </div>
        """
    )

    with gr.Row():

        # ----------------------------------------------------
        # LEFT SIDE — INPUT
        # ----------------------------------------------------

        with gr.Column(
            scale=1
        ):

            gr.Markdown(
                "## 👥 Group Details"
            )

            traveler_count = gr.Slider(
                minimum=1,
                maximum=4,
                value=4,
                step=1,
                label="Number of Travelers"
            )

            traveler1 = gr.Textbox(
                value="Beach, Historical",
                label="Traveler 1 Preferences",
                placeholder="Example: Beach, Nature"
            )

            traveler2 = gr.Textbox(
                value="Nature, Adventure",
                label="Traveler 2 Preferences"
            )

            traveler3 = gr.Textbox(
                value="City, Historical",
                label="Traveler 3 Preferences"
            )

            traveler4 = gr.Textbox(
                value="Nature, Adventure",
                label="Traveler 4 Preferences"
            )

            budget = gr.Number(
                value=5000,
                minimum=0,
                label="Accommodation Budget per Person (₹)"
            )

            accommodation_type = gr.Dropdown(
                choices=[
                    "Hotel",
                    "GuestHouse",
                    "Homestay",
                    "Luxury Camps",
                    "Boutique Hotel"
                ],
                value="Hotel",
                label="Accommodation Type"
            )

            recommend_button = gr.Button(
                "🔍 Get PACKVOTE Recommendations",
                variant="primary"
            )

            gr.Markdown(
                """
                ### 🤖 ML Pipeline

                **K-Means** → Traveler Profiling
                **KNN** → Experience / Similarity
                **Decision Tree** → Suitability
                **Naive Bayes** → Review Classification
                **Linear Regression** → Cost Prediction
                """
            )

        # ----------------------------------------------------
        # RIGHT SIDE — RESULTS
        # ----------------------------------------------------

        with gr.Column(
            scale=2
        ):

            recommendations_html = gr.HTML(
                label="Recommendations"
            )

            recommendation_table = gr.Dataframe(
                headers=None,
                interactive=False,
                label="Ranking"
            )

            tourist_html = gr.HTML(
                label="Tourist Spots"
            )

            download_file = gr.File(
                label="Download Recommendation CSV"
            )

    # --------------------------------------------------------
    # Button action
    # --------------------------------------------------------

    recommend_button.click(
        fn=run_packvote,
        inputs=[
            traveler_count,
            traveler1,
            traveler2,
            traveler3,
            traveler4,
            budget,
            accommodation_type
        ],
        outputs=[
            recommendations_html,
            recommendation_table,
            tourist_html,
            download_file
        ]
    )

# ------------------------------------------------------------
# 10. Launch
# ------------------------------------------------------------

print("✅ PACKVOTE UI created successfully")
print("Launching the application...")

demo.launch(
    share=True,
    debug=True
)
```
